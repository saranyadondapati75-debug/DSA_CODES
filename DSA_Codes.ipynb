```python
#write a code to print the sum of digits of a given number
n=int(input("Enter a number:"))
sum=0
while n!=0:
    d=n%10
    sum+=d
    n//=10
print("sum of digits",sum)

```

```python
#reverse of a given number
n=int(input("Enter a number:"))
rev=0
while n>0:
    d=n%10
    rev=rev*10+d
    n//=10
print("reverse number",rev)
```

```python
#write a code to print even and odd digits of a given number
num=int(input("enter a number"))
even=0
odd=0
while num!=0:
  d= num%10
  if d%2==0:
    even+=1
  else:
    odd+=1
  num//=10
print("Even count:" ,even)
print("odd count:", odd)
```

```python
#write a code to tom-riddle anagram
s1=input("Enter 1st string:").replace(' ','').lower()
s2=input("Enter 2st string:").replace(' ','').lower()
if len(s1)!=len(s2):
  print("Not anagram")
elif sorted(s1)==sorted(s2):
  print("Anagram")
else:
  print("not anagram.....")
```
