```python
#2D Data structure
#hallow square(odd numbers)
n=int(input("Enter a number:"))
for i in range(n):
  for j in range(n):
    if i==0 or i==n-1 or j==0 or j==n-1 or i==j or (i+j)==n-1:
      print("*",end=' ')
    else:
      print(' ',end=' ')
  print()
```

```python
#pyramid
n=int(input("Enter a number:"))
for i in range(n):
  for j in range(2*n-1):
    if i == n - 1 or j == n - 1 - i or j == n - 1 + i:
      print("*",end=' ')
    else:
      print(' ',end=' ')
  print()
```

```python
#pyramid or equilateral
n=int(input("Enter a number:"))
for i in range(1,n+1):
  for j in range(n-i):
    print(" ",end=' ')
  for j in range(2*i-1):
    if i==n or j==0 or j==2*i-2:
      print("*",end=' ')
    else:
      print(' ',end=' ')
  print()
```

```python
#inverted pyramid
n=int(input("Enter a number:"))
for i in range(n,0,-1):
  for j in range(n-i):
    print(" ",end=' ')
  for j in range(2*i-1):
    if i==n or i==1 or j==0 or j==2*i-2:
      print("*",end=' ')
    else:
      print(' ',end=' ')
  print()

```

```python
#Butterfly pattern
n=int(input("Enter value:"))
for i in range(n):
  for j in range(n):
    if j==0 or j==n-1 or i==j or i+j==n-1:
      print("*",end=" ")
    else:
      print(" ",end=" ")
  print()
```

```python
#Diamond pattern
n=int(input("Enter a number:"))
for i in range(1,n+1):
  for j in range(n-i):
    print(" ",end=' ')
  for j in range(2*i-1):
    if j==0 or j==2*i-2:
      print("*",end=' ')
    else:
      print(' ',end=' ')
  print()
for i in range(n-1,0,-1):
  for j in range(n-i):
    print(" ",end=' ')
  for j in range(2*i-1):
    if j==0 or j==2*i-2:
      print("*",end=' ')
    else:
      print(' ',end=' ')
  print()
```
