```python
#Average in an array
n=int(input("Enter size of array:"))
arr=[]
sum=0
for i in range(n):
  num=float(input("Enter a value:"))
  arr.append(num)

print("Array elements:")
for i in range(n):
  sum=sum+arr[i]
print('avg:',float(sum/n))
```

```python
#code to read n values and traverse those values and print the odd and even numbers
#arr=[12,3,5,7,9,2]
#evencount=2
#oddcount=4
n=int(input("Enter size of array:"))
arr=[]
evencount, oddcount =0,0
for i in range(n):
  num=int(input("Enter a value:"))
  arr.append(num)
for i in range(n):
  if arr[i]%2==0:
    evencount+=1
  else:
    oddcount+=1
print("EvenCount:",evencount)
print("OddCount:",oddcount)
```

```python
#code to read n values and traverse those values and print the +ve and -ve numbers
n=int(input("Enter size of array:"))
arr=[]
positivecount=0
negativecount=0
for i in range(n):
  num=int(input("Enter a value:"))
  arr.append(num)
for i in range(n):
  if arr[i]>0:
    positivecount+=1
  else:
    negativecount+=1
print("positivecount:",positivecount)
print("negativecount:",negativecount)
```
