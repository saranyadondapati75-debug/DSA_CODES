```python

#Searching types
#1.linear search
#2.Binary search
#3.Jump search

#linear search(unsorted data)
arr=list(map(int,input("Enter elements:").split()))
target=int(input("Enter a target:"))
for i in range(len(arr)):
  if arr[i]==target:
    print(target,"Found at index: ",i)
    break
```

```python
##Binary search
arr=list(map(int,input("Enter elements:").split()))
target=int(input("Enter a target:"))
left=0
right=len(arr)-1
found=False
while left<right:
  middle=(left+right)//2
  if arr[middle]==target:
    print("target found at index",middle)
    found=True
    break
  elif target< arr[middle]:
    right=middle-1
  else:
    left=middle-1
if not found:
  print("Element doesn't exist ")



```

```python
#jump search
import math
arr=list(map(int,input("Enter elements:").split()))
key=int(input("Enter a target:"))
n=len(arr)
step=int(math.sqrt(n))
prev=0
found=False
while prev<n and arr[min(step,n)-1]<key:
  prev=step
  step+=int(math.sqrt(n))
  if prev>=n:
    break
while prev<min(step,n):
  if arr[prev]==key:
    print("Element found at",prev)
    found=True
    break
  prev+=1
if not found:
  print("Element not found ")
```
